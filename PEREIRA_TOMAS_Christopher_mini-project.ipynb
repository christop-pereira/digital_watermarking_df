```python
def adjust_brightness_contrast(img, brightness=0, contrast=1.0):
    return cv2.convertScaleAbs(img, alpha=contrast, beta=brightness)
```

```python
import cv2
import numpy as np
from skimage.metrics import structural_similarity as ssim


original = cv2.imread('invisible-watermark/test_vectors/peppers.png')
watermarked_images = ['invisible-watermark/test_vectors/wm.png', 
                      'invisible-watermark/test_vectors/wm_svd.png', 
                      'invisible-watermark/test_vectors/wm_rivagan.png']

for wm in watermarked_images:
    watermarked = cv2.imread(wm)

    # PSNR Calcul
    mse = np.mean((original - watermarked) ** 2)
    psnr = 10 * np.log10((255 ** 2) / mse)
    print(f'{wm} - PSNR: {psnr:.2f} dB')

    # SSIM Calcul
    s = ssim(original, watermarked, channel_axis=2)
    print(f'{wm} - SSIM: {s:.4f}\n')
```

```python
import numpy as np

# ----------------------------
# Utility functions
# ----------------------------

def str_to_bits(s):
    """Convert string to 8-bit binary array."""
    return np.array([int(b) for char in s for b in format(ord(char), '08b')])

def ber(original, extracted):
    """Bit Error Rate."""
    if original is None or extracted is None or len(original) == 0:
        return None
    L = min(len(original), len(extracted))
    return np.sum(original[:L] != extracted[:L]) / L

def ncc(original, extracted):
    """Normalized Cross-Correlation."""
    if original is None or extracted is None:
        return None
    L = min(len(original), len(extracted))
    o = original[:L].astype(np.float32)
    e = extracted[:L].astype(np.float32)
    return float(np.sum(o * e) / np.sqrt(np.sum(o**2) * np.sum(e**2)))

# ----------------------------
# Input: your experiment results
# ----------------------------

results = [
    # ==== BASE ====
    ("Base", "DCT+DWT", "hello", "hello"),
    ("Base", "DCT+DWT+SVD", "hello", "hello"),
    ("Base", "RivaGAN", "hell", "h%ll"),

    # ==== JPEG 90 ====
    ("JPEG90", "DCT+DWT", "hello", "p"),
    ("JPEG90", "DCT+DWT+SVD", "hello", "hello"),
    ("JPEG90", "RivaGAN", "hell", "h%,l"),

    # ==== JPEG 70 ====
    ("JPEG70", "DCT+DWT", "hello", "`"),
    ("JPEG70", "DCT+DWT+SVD", "hello", "hello"),
    ("JPEG70", "RivaGAN", "hell", "h%,l"),

    # ==== JPEG 50 ====
    ("JPEG50", "DCT+DWT", "hello", ""),
    ("JPEG50", "DCT+DWT+SVD", "hello", ""),
    ("JPEG50", "RivaGAN", "hell", "h,l"),

    # ==== NOISE 20 ====
    ("Noise20", "DCT+DWT", "hello", ""),
    ("Noise20", "DCT+DWT+SVD", "hello", "H H,"),
    ("Noise20", "RivaGAN", "hell", "hhl"),

    # ==== NOISE 10 ====
    ("Noise10", "DCT+DWT", "hello", ""),
    ("Noise10", "DCT+DWT+SVD", "hello", "hello"),
    ("Noise10", "RivaGAN", "hell", "h%ll"),

    # ==== NOISE 5 ====
    ("Noise5", "DCT+DWT", "hello", "hel|0"),
    ("Noise5", "DCT+DWT+SVD", "hello", "hello"),
    ("Noise5", "RivaGAN", "hell", "h%ll"),

    # ==== Brightness / Contrast ====
    ("BrightnessContrast-50-0.8", "DCT+DWT", "hello", "Hello"),
    ("BrightnessContrast-50-0.8", "DCT+DWT+SVD", "hello", ""),
    ("BrightnessContrast-50-0.8", "RivaGAN", "hell", ""),

    ("BrightnessContrast50-1.2", "DCT+DWT", "hello", ""),
    ("BrightnessContrast50-1.2", "DCT+DWT+SVD", "hello", ""),
    ("BrightnessContrast50-1.2", "RivaGAN", "hell", "hE,l"),

    # ==== Blurring ====
    ("Blurring", "DCT+DWT", "hello", "hell/"),
    ("Blurring", "DCT+DWT+SVD", "hello", "hello"),
    ("Blurring", "RivaGAN", "hell", "h%ll"),

    # ==== Geometric scaled ====
    ("GeometricScaled", "DCT+DWT", "hello", ""),
    ("GeometricScaled", "DCT+DWT+SVD", "hello", ""),
    ("GeometricScaled", "RivaGAN", "hell", "h,L"),

    # ==== Geometric rotated ====
    ("GeometricRotated", "DCT+DWT", "hello", ""),
    ("GeometricRotated", "DCT+DWT+SVD", "hello", "127ms"),
    ("GeometricRotated", "RivaGAN", "hell", "helD"),

    # ==== Color space ====
    ("ColorSpace", "DCT+DWT", "hello", ""),
    ("ColorSpace", "DCT+DWT+SVD", "hello", ""),
    ("ColorSpace", "RivaGAN", "hello", "@"),
]

# ----------------------------
# Compute & display metrics
# ----------------------------

print("\n=== ROBUSTNESS METRICS ===\n")

for attack, method, original, extracted in results:
    print(f"\n--- {attack} | {method} ---")
    print(f"Original :  '{original}'")
    print(f"Extracted : '{extracted}'")

    # convert to bits
    orig_bits = str_to_bits(original)
    ext_bits = str_to_bits(extracted) if extracted else None

    # BER
    ber_val = ber(orig_bits, ext_bits)
    print(f"BER : {ber_val if ber_val is not None else 'N/A'}")

    # NCC
    ncc_val = ncc(orig_bits, ext_bits)
    print(f"NCC : {ncc_val if ncc_val is not None else 'N/A'}")

    # RivaGAN detection confidence (proxy)
    if "RivaGAN" in method:
        conf = 0 if ncc_val is None else max(0, min(1, (ncc_val + 1) / 2))
        print(f"Detection confidence (RivaGAN): {conf}")

print("\n=== END ===\n")
```
